```python
from selenium import webdriver #웹서버에 데이터를 요청하는 라이브러리 
import time
from tqdm import tqdm_notebook
import pandas as pd
```

```python
driver.get(url)
```

```python
html=driver.page_source
soup=driver.page_source
soup=BeautifulSoup(html,'lxml')
soup
```

```python
#포항
```

```python
name_list=soup.find_all("strong","PlaceListTitle_text__2511B normal")
name_text_list=[name.text for name in name_list]
name_text_list
```

```python
len(name_text_list)
```

```python
rename_list=[]
exp_list=[]
for i in range(42):
    rename_list.append(name_text_list[i].split('(')[0])
    if len(rename_list[i])==len(name_text_list[i]):
        exp_list.append('없음')
    else:
        exp_list.append(name_text_list[i].split('(')[1])
```

```python
reexp_list=[]
for i in range(42):
    reexp_list.append(exp_list[i].split(')')[0])
reexp_list
```

```python
point_list=soup.find_all("span","PlaceListScore_rating__3Glxf")
point_text_list=[float(point.text) for point in point_list]
point_text_list
```

```python
len(point_text_list)
```

```python
point_text_list=point_text_list[:42]
len(point_text_list)
```

```python
hotel_df=pd.DataFrame({'이름':rename_list})
hotel_df['평균평점']=point_text_list
hotel_df['설명']=reexp_list
hotel_df
```

```python
address=soup.find("span","PlaceListTitle_normal__318s-")
hotel_df['주소']=address.text
hotel_df
```
